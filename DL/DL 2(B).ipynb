##2. Binary classification using Deep Neural Networks Example: Classify movie reviews into  positive" reviews and "negative" reviews, just based on the text content of the reviews.  Use IMDB dataset 

```python
# =========================
# 1. IMPORT LIBRARIES
# =========================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from tensorflow import keras
from tensorflow.keras import layers
```

```python
# Step 2: Load IMDB Dataset
(x_train, y_train), (x_test, y_test) = keras.datasets.imdb.load_data(num_words=10000)
```

```python
# Create DataFrame
df = pd.DataFrame({
    "Review": x_train,
    "Sentiment": y_train
})

# First 5 Rows
print("First 5 Rows:\n")
print(df.head())

# Shape of Dataset
print("\nDataset Shape:")
print(df.shape)

# Dataset Information
print("\nDataset Info:")
print(df.info())

# Column Names
print("\nColumn Names:")
print(df.columns)

# Sentiment Count
print("\nSentiment Counts:")
print(df["Sentiment"].value_counts())
```

```python
# Step 3: Convert to same length (padding)
x_train = keras.preprocessing.sequence.pad_sequences(x_train, maxlen=200)
x_test = keras.preprocessing.sequence.pad_sequences(x_test, maxlen=200)
```

```python
# Step 4: Build Model
model = keras.Sequential([
    layers.Embedding(10000, 32, input_length=200),
    layers.Flatten(),
    layers.Dense(32, activation='relu'),
    layers.Dense(1, activation='sigmoid')   # Binary output
])
```

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

```python
# Step 6: Train Model
model.fit(x_train, y_train, epochs=5, batch_size=32, validation_split=0.2)
```

```python
# Step 7: Evaluate Model
loss, acc = model.evaluate(x_test, y_test)
print("Test Accuracy:", acc)
```

```python
# Prediction on full test data
pred = model.predict(x_test)

# Convert probabilities to 0/1
y_pred = (pred > 0.5).astype("int32")
```

```python
# =========================
# ACTUAL VS PREDICTED
# =========================

df_test = pd.DataFrame({
    "Review": [str(review) for review in x_test[:10]],
    "Actual": y_test[:10],
    "Predicted": y_pred[:10].flatten()
})

print(df_test)
```
